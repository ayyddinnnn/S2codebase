**case** Find the gradient of the scalar field $f=x^2+y^2+z^2& and plot the gradient field


**solution**

Evaluating gradient of a scalar field

```python
import sympy as sp
#Define the symbols
x,y,z = sp.symbols(' x y z')

#Define the scalar field
scalar_field = x**2 + y**2 + z**2 #Example scalar field
scalar_field
```

```python
from random import getrandbits
#Compute the gradient
gradiant= sp.Matrix([sp.diff(scalar_field , var) for var in (x,y,z)])
print("the gradient of the scalar field is:")
gradiant
```

```python
import numpy as np
import matplotlib.pyplot as plt

#Define a grid of points where you want to evaluate the gradient

x_vals=np.linspace(-10,10,20)
y_vals=np.linspace(-10,10,20)
z_vals=0

X,Y=np.meshgrid(x_vals,y_vals)

#Initialise the arrays to store gradient components
U= np.zeros_like(X, dtype=float)
V=np.zeros_like(Y, dtype=float)
#Evaluate the gradient at each point
for i in range(X.shape[0]):
  for j in range(X.shape[1]):
    grad_evaluated=gradiant.subs({x:X[i,j],y:Y[i,j],z:z_vals})
    U[i,j]=grad_evaluated[0]
    V[i,j]=grad_evaluated[1]
#Plot the gradient vectors using matplotlib
fig,ax= plt.subplots()
ax.quiver(X,Y,U,V)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Gradient Vectors of the scalar field")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Define the region of the plate
x_vals= np.linspace(-5,5,100)
y_vals = np.linspace(-5,5,100)
X, Y= np.meshgrid(x_vals, y_vals)

#Define the temperature distribution function
T=100*np.exp(-(X**2+Y**2)/2)

#Create a contour plot of the temperature distribution
plt.contourf(X, Y, T, levels=10, cmap='inferno')
plt.colorbar(label='Temperature(c)')
plt.title('temperature distribution T (x,y)')
plt.xlabel('x (meters)')
plt.ylabel('y (meters)')
plt.show()
```

```python
# Re-initialize U and V with the correct dimensions based on current X and Y
U = np.zeros_like(X, dtype=float)
V = np.zeros_like(Y, dtype=float)

# Evaluate the gradient at each point using the correct gradient 'grad'
# Note: scalar_field1 is 2D, so 'z' is not relevant for substitution
for i in range(X.shape[0]):
  for j in range (X.shape[1]):
    # Use 'grad' instead of 'gradient' (which was for x^2+y^2+z^2)
    # and only substitute x and y, as 'grad' is a 2D gradient.
    grad_evaluated = grad.subs({x: X[i, j], y: Y[i, j]})
    U[i, j] = float(grad_evaluated[0])
    V[i, j] = float(grad_evaluated[1])

# Plot the gradient vectors using matplotlib
fig, ax = plt.subplots()
ax.quiver(X, Y, U, V)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Gradient of the scalar field (100*exp(-(x**2+y**2)/2))')

plt.show()
```
